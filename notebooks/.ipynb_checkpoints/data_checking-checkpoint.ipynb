```python
month_name_num = {
    "Jan": "01",
    "Feb": "02",
    "Mar": "03",
    "Apr": "04",
    "May": "05",
    "Jun": "06",
    "Jul": "07",
    "Aug": "08",
    "Sep": "09",
    "Oct": "10",
    "Nov": "11",
    "Dec": "12",
}
```

```python
month_dfs = {}
```

```python
for key, value in month_name_num.items():
    df = pd.read_parquet(f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2023-{value}.parquet")
    cleaned_df = clean_data(df, int(value))
    month_dfs[key] = cleaned_df
```

```python
df_feb = month_dfs["Feb"][0]
df_mar = month_dfs["Mar"][0]
```

```python
pd.concat([df_feb, df_mar])
```
